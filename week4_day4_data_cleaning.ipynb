```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset correctly
df = pd.read_csv(r'C:\Users\anjan\ML-Journey\adult\adult.data', 
                header=None, 
                na_values='?')

# Assign proper column names
column_names = [
    'age', 'workclass', 'fnlwgt', 'education', 'education-num',
    'marital-status', 'occupation', 'relationship', 'race', 'sex',
    'capital-gain', 'capital-loss', 'hours-per-week', 'native-country', 'income'
]

df.columns = column_names

print("=== Dataset Loaded Successfully ===")
print("Shape of dataset:", df.shape)

print("\n=== First 5 Rows ===")
print(df.head())

print("\n=== Missing Values in Each Column ===")
print(df.isnull().sum())

print("\n=== Percentage of Missing Values ===")
missing_percent = (df.isnull().sum() / len(df) * 100).round(2)
print(missing_percent)

#check for duplicate rows
print("\n number of duplicate rows")
print(df.duplicated().sum())

#check data types of all columns
print("\n data types of each column")
print(df.dtypes)

#basic statistical summary of numerical columns
print("\n statistical summary")
print(df.describe())

#remove duplicate rows
print("number of rows removing duplicate:",len(df))
df = df.drop_duplicates()
print("number of rows after removing duplicates:",len(df))

#check value counts of important categories columns
print(" workclass distribution")
print(df['workclass'].value_counts())

print("\n education distribution")
print(df['education'].value_counts())

print("\n marital status")
print(df['marital-status'].value_counts())

print("\n occupation distribution")
print(df['occupation'].value_counts())

print("\n race distribution")
print(df['race'].value_counts())

print("\n sex distribution")
print(df['sex'].value_counts())

print("\n income distribution")
print(df['income'].value_counts())

print("\n native country distribution")
print(df['native-country'].value_counts().head(10))

#replace "?" with unknown in cols that have missing values
df['workclass']=df['workclass'].replace(' ?','unknown')
df['occupation']=df['occupation'].replace(' ?','unknown')
df['native-country'] = df['native-country'].replace(' ?','unknown')

print("missing values after replacement")
print(df.isnull().sum())
print("\n workclass value counts after cleaning")
print(df['workclass'].value_counts())


print("Unique values in workclass:")
print(df['workclass'].unique())

# Save the cleaned dataset
df.to_csv('adult_cleaned.csv', index=False)

print("Cleaned dataset saved successfully as 'adult_cleaned.csv'")
print("Final shape of cleaned data:", df.shape)

# set professional style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize']=(9,6)

#1.income distribution(target variable)
plt.figure()
sns.countplot(data=df, x='income', palette='Set2')
plt.title('Income Distribution (<=50K vs >50K)')
plt.xlabel('Income')
plt.ylabel('Count')
plt.show()

# 2. Education Level vs Income
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='education', hue='income', order=df['education'].value_counts().index)
plt.title('Education Level vs Income')
plt.xticks(rotation=45)
plt.xlabel('Education')
plt.ylabel('Count')
plt.legend(title='Income')
plt.show()

# 3. Age Distribution by Income
plt.figure()
sns.boxplot(data=df, x='income', y='age', palette='coolwarm')
plt.title('Age Distribution by Income Level')
plt.xlabel('Income')
plt.ylabel('Age')
plt.show()

# 4. Hours per Week by Income
plt.figure()
sns.boxplot(data=df, x='income', y='hours-per-week', palette='viridis')
plt.title('Hours Worked per Week by Income Level')
plt.xlabel('Income')
plt.ylabel('Hours per Week')
plt.show()

# 5. Gender vs Income
plt.figure()
sns.countplot(data=df, x='sex', hue='income', palette='Set1')
plt.title('Gender vs Income Level')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.legend(title='Income')
plt.show()
```
