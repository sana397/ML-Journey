```python
import numpy as np
X=np.array([1,2,3,4,5])
y=np.array([2,4,5,4,5])

#add a column of ones X
X_b=np.c_[np.ones((len(X),1)),X]
print(X_b)

#normal equation
theta_best=np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y
print(theta_best)
y_pred=X_b @ theta_best
print(y)
print(y_pred)

#calculate MSE
mse=np.mean((y - y_pred)**2)
print("MSE:",mse)

alpha=theta_best[0]
beta=theta_best[1]
print("intercept(aplha):",alpha)
print("slope(beta):",beta)

```
