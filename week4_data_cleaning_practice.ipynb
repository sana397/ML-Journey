```python
import pandas as pd
import numpy as np

# Create messy data
data = {
    'emp_id': [101, 102, 103, 104, 105, 106, 107, 102],
    'name': ['anjani', 'jahnavi', 'nithya', 'bhavani', 'rahul', None, 'sneha', 'jahnavi'],
    'sal': [70000, 55000, 90000, '75000', 80000, 62000, None, 55000],
    'dept': ['IT', 'HR', 'FINANCE', 'IT', 'HR', 'FINANCE', 'IT', 'HR'],
    'Joining_Date': ['2020-05-12', '2019-03-15', '2021-07-20', '2018-11-05', 
                     '2022-01-10', '2020-09-30', '2017-06-18', '2019-03-15'],
    'Performance_Score': [85, 92, None, 78, 88, 95, 70, 92]
}

df = pd.DataFrame(data)
print("=== Original Messy Data ===")
print(df)
print("\nOriginal Data Types:\n", df.dtypes)

# Handling Missing Values
print("\n=== Missing Values Count ===")
print(df.isnull().sum())

# Fill missing values
df['name'] = df['name'].fillna('unknown')
df['sal'] = df['sal'].fillna(df['sal'].median())
df['Performance_Score'] = df['Performance_Score'].fillna(df['Performance_Score'].mean())

print("\n=== After Filling Missing Values ===")
print(df)

# Remove Duplicates
print("\n=== Duplicate Rows ===")
print(df.duplicated().sum())

df = df.drop_duplicates()
print("\n=== After Removing Duplicates ===")
print(df)

# Fix Wrong Data Types (with error handling)
print("\n=== Fixing Data Types ===")

df['sal'] = pd.to_numeric(df['sal'], errors='coerce')
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'], errors='coerce')

print("\nUpdated Data Types:\n", df.dtypes)
print(df)

# Rename Columns (Correct column names)
df = df.rename(columns={
    'Joining_Date': 'Join_Date',
    'Performance_Score': 'Perf_Score'
})

print("\n=== After Renaming Columns ===")
print(df.columns.tolist())
print(df)

# Final Cleaned Data
print("\n=== Final Cleaned Data ===")
print(df)
print("\nMissing Values After Cleaning:\n", df.isnull().sum())
print("\nShape of Clean Data:", df.shape)



```
