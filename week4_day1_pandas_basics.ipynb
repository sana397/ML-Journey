```python
import pandas as pd
import numpy as np
#basic visualisation
import matplotlib.pyplot as plt
import seaborn as sns

data={'patient_id': [1, 2, 3, 4, 5],
    'age': [65, 72, 58, 80, 45],
    'gender': ['M', 'F', 'M', 'M', 'F'],
    'label': ['Normal', 'AFib', 'Normal', 'AFib', 'Normal'],
    'heart_rate': [72, 118, 68, 142, 75],
    'rr_interval': [0.83, 0.51, 0.88, 0.42, 0.80]}

df=pd.DataFrame(data)
print(df)

print("\n shape of dataframe:",df.shape)

print("columns:",df.columns.tolist())

print("datatypes:\n",df.dtypes)

print("\n basic inspection")
print(df.head(3))#prints first 3 rows 
print(df.info())#prints summary
print(df.describe())#prints statistical summary


print("\n selecting data")
#selecting single column
print(df['age'])

#selecting multiple columns
print(df[['age','label','heart_rate']])

#selecting rows using index
print(df.iloc[0:3])#first three rows by position
#filter based on condition 
print("\n patients with AFib:")
print(df[df['label']=='AFib'])
print("\n patients older than 60:")
print(df[df['age']>60])

# Count Plot - Class Distribution
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='label')
plt.title('Distribution of Normal vs AFib')
plt.show()

# Histogram - Age Distribution
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='age', kde=True)
plt.title('Age Distribution')
plt.show()

# Boxplot - Heart Rate by Label
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='label', y='heart_rate')
plt.title('Heart Rate Distribution: Normal vs AFib')
plt.show()



```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = {'patient_id': range(1, 11),
    'age': [65, 72, 58, 80, 45, 67, 55, 78, 62, 70],
    'gender': ['M', 'F', 'M', 'M', 'F', 'F', 'M', 'M', 'F', 'M'],
    'label': ['Normal', 'AFib', 'Normal', 'AFib', 'Normal', 'AFib', 'Normal', 'Normal', 'AFib', 'Normal'],
    'heart_rate': [72, 118, 68, 142, 75, 130, 82, 95, 125, 88],
    'rr_std': [0.08, 0.25, 0.06, 0.30, 0.07, 0.22, 0.09, 0.11, 0.28, 0.10]
}

df=pd.DataFrame(data)
print(" dataset created sucessfully")
print("1.shape:",df.shape)
print("\n 2.first 5 rows :",df.head())
print("\n 3.last 3 rows :",df.tail(3))
print("\n 4.columns names:",df.columns.tolist())
print("\n 5.datatypes :\n",df.dtypes)

# Selecting columns
print("\n6. Selecting specific columns:")
print(df[['age', 'label', 'heart_rate']])

# Filtering
print("\n7. Patients with AFib:")
print(df[df['label'] == 'AFib'])

print("\n8. Patients older than 65:")
print(df[df['age'] > 65])

# Statistics
print("\n9. Average heart rate by label:")
print(df.groupby('label')['heart_rate'].mean())

print("\n10. Summary statistics:")
print(df.describe())
# 1. Count Plot
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='label')
plt.title('Number of Normal vs AFib Cases')
plt.show()

# 2. Scatter Plot
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='age', y='heart_rate', hue='label')
plt.title('Age vs Heart Rate')
plt.show()

# 3. Box Plot
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='label', y='heart_rate')
plt.title('Heart Rate Distribution by Class')
plt.show()

```

```python
import wfdb
import os
import matplotlib.pyplot as plt
import pandas as pd

os.makedirs('../data', exist_ok=True)

print("Trying to load a sample ECG record...")

try:
    # Using a known valid record from MIT-BIH AF Database (more reliable for beginners)
    record = wfdb.rdrecord('04015', 
                           pn_dir='afdb/1.0.0/', 
                           sampfrom=0, 
                           sampto=10000)
    
    print("✅ Successfully loaded real ECG data!")
    print("Record Name:", record.record_name)
    print("Signal Shape:", record.p_signal.shape)
    print("Sampling Rate:", record.fs, "Hz")
    print("Channels:", record.sig_name)
    
except Exception as e:
    print("❌ Error:", str(e))
    print("\nWe'll use dummy data as fallback for now.")
# Plot First Channel (ECG1)
plt.figure(figsize=(14, 6))
plt.plot(record.p_signal[:, 0], color='blue', linewidth=1.2, label='ECG Channel 1')
plt.title('Real ECG Signal - Record 04015 (First 10 seconds)')
plt.xlabel('Sample Number')
plt.ylabel('Amplitude (mV)')
plt.grid(True)
plt.legend()
plt.show()

# Plot Second Channel (ECG2) for comparison
plt.figure(figsize=(14, 6))
plt.plot(record.p_signal[:, 1], color='red', linewidth=1.2, label='ECG Channel 2')
plt.title('Second ECG Channel - Record 04015')
plt.xlabel('Sample Number')
plt.ylabel('Amplitude (mV)')
plt.grid(True)
plt.legend()
plt.show()

# Basic Statistics using Pandas
ecg1 = pd.Series(record.p_signal[:, 0])
print("ECG Channel 1 Statistics:")
print(ecg1.describe())

```
